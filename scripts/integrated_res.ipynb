## Combined Results - All Models


```python
# Collect all results from different models
all_results = []

# MLP Results
file_path_mlp = 'checkpoints_mlp/'
if os.path.exists(file_path_mlp):
    df_list_mlp = []
    for file in os.listdir(file_path_mlp):
        if '.csv' in file:
            df_temp = pd.read_csv(file_path_mlp + file, index_col=0)
            df_temp['data'] = file.replace('_best_params_res.csv', '')
            df_temp['model_type'] = 'MLP'
            df_list_mlp.append(df_temp)
    if df_list_mlp:
        all_results.extend(df_list_mlp)

# LSTM Results
file_path_lstm = 'checkpoints_lstm/'
if os.path.exists(file_path_lstm):
    df_list_lstm = []
    for file in os.listdir(file_path_lstm):
        if '.csv' in file:
            df_temp = pd.read_csv(file_path_lstm + file, index_col=0)
            df_temp['data'] = file.replace('_best_params_res.csv', '')
            df_temp['model_type'] = 'LSTM'
            df_list_lstm.append(df_temp)
    if df_list_lstm:
        all_results.extend(df_list_lstm)

# Transformer Results
file_path_transformer = 'checkpoints_transformer/'
if os.path.exists(file_path_transformer):
    df_list_transformer = []
    for file in os.listdir(file_path_transformer):
        if '.csv' in file:
            df_temp = pd.read_csv(file_path_transformer + file, index_col=0)
            df_temp['data'] = file.replace('_best_params_res.csv', '')
            df_temp['model_type'] = 'Transformer'
            df_list_transformer.append(df_temp)
    if df_list_transformer:
        all_results.extend(df_list_transformer)

# Improved MLP Results
file_path_improved_mlp = '../checkpoints_improved_mlp/'
if os.path.exists(file_path_improved_mlp):
    df_list_improved_mlp = []
    for file in os.listdir(file_path_improved_mlp):
        if '.csv' in file:
            df_temp = pd.read_csv(file_path_improved_mlp + file, index_col=0)
            df_temp['data'] = file.replace('_best_params_res.csv', '')
            df_temp['model_type'] = 'Improved_MLP'
            df_list_improved_mlp.append(df_temp)
    if df_list_improved_mlp:
        all_results.extend(df_list_improved_mlp)

print(f"Total result files found: {len(all_results)}")
```

```python
if all_results:
    # Combine all results
    df_all_results = pd.concat(all_results, ignore_index=True)
    
    # Clean metric columns (remove tensor() wrapper if present)
    metric_cols = ['final model mae', 'final model mse', 'final model rmse', 'final model mape',
                   'best val model mae', 'best val model mse', 'best val model rmse', 'best val model mape']
    
    for col in metric_cols:
        if col in df_all_results.columns:
            df_all_results[col] = df_all_results[col].apply(
                lambda x: str(x).replace(')', '').replace('tensor(', '').strip() if pd.notna(x) else x
            )
            # Convert to numeric if possible
            try:
                df_all_results[col] = pd.to_numeric(df_all_results[col], errors='ignore')
            except:
                pass
    
    # Sort by model_type and data
    df_all_results = df_all_results.sort_values(['model_type', 'data'])
    
    print(f"Combined results shape: {df_all_results.shape}")
    print(f"\nModel types: {df_all_results['model_type'].unique()}")
    print(f"\nData files: {df_all_results['data'].unique()}")
    
    df_all_results
else:
    print("No results found. Please run the models first.")
```

```python
# Export combined results to Excel
if all_results and 'df_all_results' in locals():
    output_file = 'all_models_integrated_results.xlsx'
    df_all_results.to_excel(output_file, index=False)
    print(f"✅ Combined results exported to: {output_file}")
    print(f"   Total rows: {len(df_all_results)}")
    print(f"   Total columns: {len(df_all_results.columns)}")
else:
    print("No results to export.")
```

## Summary Statistics by Model Type


```python
# Create summary statistics by model type
if all_results and 'df_all_results' in locals():
    summary_cols = ['final model mae', 'final model mse', 'final model rmse', 'final model mape']
    available_cols = [col for col in summary_cols if col in df_all_results.columns]
    
    if available_cols:
        summary_stats = df_all_results.groupby('model_type')[available_cols].agg(['mean', 'std', 'min', 'max'])
        print("Summary Statistics by Model Type:")
        print(summary_stats)
        
        # Export summary to Excel
        summary_file = 'model_comparison_summary.xlsx'
        summary_stats.to_excel(summary_file)
        print(f"\n✅ Summary statistics exported to: {summary_file}")
    else:
        print("No metric columns found for summary.")
else:
    print("No results available for summary.")
```
